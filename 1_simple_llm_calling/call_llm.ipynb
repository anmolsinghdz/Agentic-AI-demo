```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
response = llm.invoke("Who is the president of USA?")

print(response.content)
```

```python
# giving system instructions to LLM
response = llm.invoke([
    ["system","You are a helpful assistant that answers in one short line"],
    ["human","Who is the president of USA?"]
])

print(response.content)
```

```python
# giving temperature parameter to LLM
prompt = "write a single tagline for a quirky coffee shop. Only output the tagline nothing else."

llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", temperature=0)
response = llm.invoke(prompt)

print(response.content)
```

```python
# giving temperature parameter to LLM
prompt = "write a single tagline for a quirky coffee shop. Only output the tagline nothing else."

llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", temperature=1)
response = llm.invoke(prompt)

print(response.content)
```

```python
# calling using Groq models
from langchain_groq import ChatGroq

llm = ChatGroq(model="qwen/qwen3.8-27b", reasoning_format="parsed")

messages = [
    ("system", "you are a helpful assistant that gives short answers to questions"),
    ("human", "How many moons does jupiter has?")
]

response = llm.invoke(messages)
print(response.content)
```
